# **Matrix Operation in List**

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

def display_matrix(mat):
    for row in mat:
        print(row)
        
display_matrix(matrix)        
```

```python
# traversal

for row in matrix:
    for element in row:
        print(element, end=" ")
        
    print()    
```

```python

for i in range(len(matrix)):
    print(matrix[i])
```

```python
matrix[1][2]
```

```python
# Updation

matrix[1][2]= 15
display_matrix(matrix)
```

```python
# insertion
matrix[0].insert(1, 77)
matrix[0]
```

```python
display_matrix(matrix)
```

```python
matrix[1].append(30)
matrix[2].append(67)
display_matrix(matrix)
```

```python
# deletion
matrix[2].remove(67)
display_matrix(matrix)
```

```python
matrix[1].pop()
matrix[0].pop()
```
