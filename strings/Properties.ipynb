# Properties of strings

## Immutable

```python
str1 = 'Hello'
print(id(str1))

str1 = str1 + ' World'
print(str1)
print(id(str1))
```

# Iterable

```python
for char in str1:
    print(char, end = '')
```

# Slicing

```python
str1
```

```python
str1[1:4]
```

```python
str1[1:9:2]
```

# unicode based

```python
str1
```

```python
for i in str1:
    print(i, ord(i))
```
